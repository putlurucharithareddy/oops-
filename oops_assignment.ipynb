```python
#Q1.1

class Employee:
    def __init__(self):
        self.EmployeeID = 0
        self.Gender = ""
        self.Salary = 0
        self.PerformanceRating = 0

    def get(self):
        self.EmployeeID = int(input("Enter Employee ID: "))
        self.Gender = input("Enter Gender: ")
        self.Salary = int(input("Enter Salary: "))
        self.PerformanceRating = int(input("Enter Performance Rating (out of 5): "))
```

```python
#Q1.2

from datetime import datetime

class JoiningDetail:
    def __init__(self):
        self.DateOfJoining = None

    def getDoJ(self):
        doj = input("Enter Date of Joining (YYYY-MM-DD): ")
        self.DateOfJoining = datetime.strptime(doj, "%Y-%m-%d").date()
```

```python
#Q1.3

class Information(Employee, JoiningDetail):
    def __init__(self):
        Employee.__init__(self)
        JoiningDetail.__init__(self)

    def readData(self):
        print("\nEmployee Details")
        print("EmployeeID:", self.EmployeeID)
        print("Gender:", self.Gender)
        print("Salary:", self.Salary)
        print("PerformanceRating:", self.PerformanceRating)
        print("DateOfJoining:", self.DateOfJoining)


# ✅ Main Program
n = int(input("Enter number of employees: "))
employees = []

for i in range(n):
    print(f"\n--- Enter details for Employee {i+1} ---")
    emp = Information()
    emp.get()
    emp.getDoJ()
    employees.append(emp)

# Top 3 based on rating
top3 = sorted(employees, key=lambda x: x.PerformanceRating, reverse=True)[:3]

# Sort top3 by joining date
top3_sorted = sorted(top3, key=lambda x: x.DateOfJoining)

print("\n✅ Top 3 Employees (By Rating, Sorted by Date of Joining):")
for emp in top3_sorted:
    emp.readData()
```

```python
#Q3

def Max_Score(M):
    total_scores = {}

    # Sum scores of each player across matches
    for match in M:
        for player, score in M[match].items():
            total_scores[player] = total_scores.get(player, 0) + score

    # Find player with maximum total score
    top_player = max(total_scores, key=total_scores.get)

    return (top_player, total_scores[top_player])


# Example input dictionary
M = {
    "test1": {"Dhoni": 56, "Balaji": 94},
    "test2": {"Balaji": 80, "Dravid": 105}
}

print("✅ Output:", Max_Score(M))
```

```python
#Q4

import random

# Create deck (1 to 13 repeated 4 times)
deck = list(range(1, 14)) * 4
random.shuffle(deck)

# Pick 8 random cards
cards = deck[:8]

score = 0
current = cards[0]

print("✅ Card Game Started!")
print("First Card:", current)

for i in range(1, len(cards)):
    choice = input("Predict next card (higher/lower): ").lower()
    next_card = cards[i]

    print("Next Card:", next_card)

    if next_card == current:
        score -= 15
        print("❌ Same value! -15 points")

    elif choice == "higher" and next_card > current:
        score += 20
        print("✅ Correct! +20 points")

    elif choice == "lower" and next_card < current:
        score += 20
        print("✅ Correct! +20 points")

    else:
        score -= 15
        print("❌ Wrong! -15 points")

    current = next_card
    print("Current Score:", score)
    print("-" * 30)

print("\n🎉 Game Over! Final Score:", score)
```

```python
#Q6

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

# Example
r = Rectangle(5, 4)
print("✅ Rectangle Area =", r.area())
```

```python
#Q7

class Parent:
    def __init__(self):
        self._protected = "Protected Member"   # protected
        self.__private = "Private Member"      # private

    def show(self):
        print("Parent show() method")

    def get_private(self):
        return self.__private


class Child(Parent):
    # polymorphism (method overriding)
    def show(self):
        print("Child show() method (Overridden)")


# Create object
c = Child()

# Polymorphism
c.show()

# Protected access
print("✅ Protected:", c._protected)

# Private access using getter
print("✅ Private:", c.get_private())
```

```python
#Q11

class Triangle:
    number_of_sides = 3

    def __init__(self, angle1, angle2, angle3):
        self.angle1 = angle1
        self.angle2 = angle2
        self.angle3 = angle3

    def check_angles(self):
        total = self.angle1 + self.angle2 + self.angle3
        if total == 180:
            print("✅ Valid Triangle (Sum = 180)")
            return True
        else:
            print("❌ Not a Triangle")
            return False


# Example
t = Triangle(60, 60, 60)
t.check_angles()
```

```python
#Q11.2

def triangle_type(self):
    if not self.check_angles():
        return

    angles = [self.angle1, self.angle2, self.angle3]

    if any(a > 90 for a in angles):
        print("✅ Obtuse Triangle")
    elif any(a == 90 for a in angles):
        print("✅ Right Triangle")
    else:
        print("✅ Acute Triangle")

# attach method to Triangle class
Triangle.triangle_type = triangle_type
```

```python
#Q11.3

t1 = Triangle(60, 60, 60)
t1.triangle_type()
```

```python
#Q11.4

t2 = Triangle(90, 45, 45)
t2.triangle_type()

t3 = Triangle(100, 40, 40)
t3.triangle_type()
```

```python
#Q12


class isosceles_triangle(Triangle):
    def property_check(self, s1, s2, s3):
        if s1 == s2 or s2 == s3 or s1 == s3:
            print("✅ Isosceles Triangle")
        else:
            print("❌ Not Isosceles Triangle")


class right_triangle(Triangle):
    def property_check(self):
        if self.angle1 == 90 or self.angle2 == 90 or self.angle3 == 90:
            print("✅ Right Triangle")
        else:
            print("❌ Not Right Triangle")


class equilateral_triangle(Triangle):
    def property_check(self):
        if self.angle1 == 60 and self.angle2 == 60 and self.angle3 == 60:
            print("✅ Equilateral Triangle")
        else:
            print("❌ Not Equilateral Triangle")
```

```python
#Q12.1

class isosceles_right_triangle(isosceles_triangle, right_triangle):
    def check_isosceles_right(self, s1, s2, s3):
        print("\n✅ Checking Isosceles Right Triangle:")

        # Isosceles check using sides
        isosceles_triangle.property_check(self, s1, s2, s3)

        # Right triangle check using angles
        right_triangle.property_check(self)


# ✅ Testing
iso = isosceles_triangle(70, 70, 40)
iso.property_check(5, 5, 8)

rt = right_triangle(90, 60, 30)
rt.property_check()

eq = equilateral_triangle(60, 60, 60)
eq.property_check()

irt = isosceles_right_triangle(90, 45, 45)
irt.check_isosceles_right(5, 5, 7)
```
